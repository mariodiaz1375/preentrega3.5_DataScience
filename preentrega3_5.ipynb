```python
import pandas as pd

# se crea el objeto df al leer el dataset en csv
df = pd.read_csv("dataset_crudo.csv")
# mostramos las 5 primeras filas
df.head(5)
```

```python
# verificamos las dimensiones del dataset con shape
print(f"Dimensiones del dataset: {df.shape}")
print("\nResumen del dataser con info():\n")
df.info()
```

```python
print("--- VALORES NULOS POR COLUMNA ---")

# calculamos la cantidad de nulos en cada columna
nulos_cant = df.isnull().sum()
# calculamos el porcentaje de nulos en cada columna
nulos_pct = df.isnull().mean() * 100

# creamos el DataFrame resumen
df_nulos = pd.DataFrame({
    'Cantidad': nulos_cant,
    'Porcentaje(%)': nulos_pct.round(3)
})

# filtramos solo las columnas con nulos mayores a 0
display(df_nulos[df_nulos['Cantidad'] > 0])
```

```python
# filtramos y mostramos todas las filas que contienen al menos un valor nulo
registros_con_nulos = df[df.isnull().any(axis=1)]

print(f"Se encontraron {len(registros_con_nulos)} registros con valores nulos:")
display(registros_con_nulos)
```

```python
# borramos los dos registros que contienen 2 nulos cada uno
df = df.drop([1500, 1502])


# para 'descuento_porcentaje' y 'descuento' reemplazamos los nulos por 0
df['descuento_porcentaje'] = df['descuento_porcentaje'].fillna(0)
df['descuento'] = df['descuento'].fillna(0)

# verificamos que ya no queden valores nulos en el df
print("--- Cantidad de nulos después de la imputación ---")
display(df.isnull().sum())
```

```python
print("--- ESTADISTICAS DESCRIPTIVAS ---")
df.describe()
```

```python
def buscar_outliers(df, columna):
  """
      esta funcion recibe un dataframe de pandas y una columna
      para devolver un dataframe con los outliers encontrados
  """
  q1 = df[columna].quantile(0.25)
  q3 = df[columna].quantile(0.75)
  iqr = q3 - q1
  outliers_cantidad1 = df[df[columna] > (q3 + 1.5 * iqr)]
  outliers_cantidad2 = df[df[columna] < (q1 - 1.5 * iqr)]
  outliers_cantidad = pd.concat([outliers_cantidad1, outliers_cantidad2])
  if outliers_cantidad.empty:
      return None
  else:
      return outliers_cantidad

# mostramos los outliers en la columna cantidad
outliers_cantidad = buscar_outliers(df, 'cantidad')
if outliers_cantidad is not None:
  print("\n--- OUTLIERS EN CANTIDAD ---")
  print(outliers_cantidad[['id_venta', 'nombre_apellido', 'producto', 'cantidad']])

# mostramos los outliers en la columna cantidad
outliers_demora = buscar_outliers(df, 'demora_entrega')
if outliers_demora is not None:
  print("\n--- OUTLIERS EN DEMORA DE ENTREGA ---")
  print(outliers_demora[['id_venta', 'nombre_apellido', 'provincia', 'demora_entrega']])
```

```python
# CONVERSION DE COLUMNAS
# decidi crear una funcion para normalizar las columnas que contienen precios
# de esta forma voy a tener preparado el DataFrame para mas adelante hacer calculos numericos

def normalizar_precios(columna):
  """
  Recibe una columna y la normaliza a float64 de forma segura,
  preservando y respetando los valores nulos (NaN) nativos.
  """
  # trabajamos solo con las filas que no sean nulas para aplicar el reemplazo de texto
  filas_no_nulas = df[columna].notna()

  # convertimos a string y limpiamos solo los valores existentes
  valores_limpios = df.loc[filas_no_nulas, columna].astype(str)
  valores_limpios = valores_limpios.str.replace('$', '', regex=False).str.strip()
  valores_limpios = valores_limpios.str.replace('.', '', regex=False)
  valores_limpios = valores_limpios.str.replace(',', '.', regex=False)

  # convertimos a float usando pd.to_numeric para mayor seguridad
  df.loc[filas_no_nulas, columna] = pd.to_numeric(valores_limpios, errors='coerce')

  # aseguramos que toda la columna sea float64 en el DataFrame
  df[columna] = df[columna].astype(float)
```

```python
# aplicamos la función normalizar_precios a las columnas monetarias
columnas_monetarias = ['precio_unitario', 'subtotal', 'descuento', 'TOTAL']

for col in columnas_monetarias:
    normalizar_precios(col)

display(df[columnas_monetarias].info())
display(df[columnas_monetarias].head())
```

```python
# convertimos a numerica la columna de descuento_porcentaje
df['descuento_porcentaje'] = df['descuento_porcentaje'].str.rstrip('%').astype(float) / 100
df['descuento_porcentaje'] = df['descuento_porcentaje'].round(2)
display(df['descuento_porcentaje'].head())
```

```python
# verificamos la cantidad de ventas por marca
ventas_por_marca = df.groupby('marca').size().reset_index(name='cantidad_ventas')

# ordenamos el resultado por la cantidad_ventas
ventas_por_marca = ventas_por_marca.sort_values(by='cantidad_ventas', ascending=False)

print("--- Ventas por marca  ---")
display(ventas_por_marca)
```

```python
ventas_por_categoria = df.groupby('categoria')['id_venta'].count().reset_index(name='cantidad_ventas')

ventas_por_categoria = ventas_por_categoria.sort_values(by='cantidad_ventas', ascending=False)

print("--- Ventas por categoria  ---")
display(ventas_por_categoria)
```

```python
# total por provincia
total_por_provincia = df.groupby('provincia')['TOTAL'].sum().reset_index(name='total_ventas')

# ordenamos el resultado por el total_ventas
total_por_provincia = total_por_provincia.sort_values(by='total_ventas', ascending=False)

print("--- Total por provincia  ---")
display(total_por_provincia)
```
